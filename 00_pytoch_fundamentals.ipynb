```python
# Matrix Multiplication
tensor_1 = torch.tensor([[1,2],[3,4]])
tensor_2 = torch.tensor([[5,6],[7,8]])
print(tensor_1.ndim)
print(tensor_2.ndim)
tensor_3 = torch.matmul(tensor_1,tensor_2)
tensor_3
```

```python
# Tensor Transpose
# Shapes need to be in the right way
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]], dtype=torch.float32)

tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]], dtype=torch.float32)
print(tensor_A.shape)
print(tensor_B.shape)

# torch.matmul(tensor_A, tensor_B) --> (this will error)
# We need to transpose
result_tensor = torch.mm(tensor_A,tensor_B.T)
result_tensor, result_tensor.shape
```

```python
# Reshaping, stacking, squeezing and unsqueezing
# Reshaping
tensor = torch.arange(1.,9.)
tensor.shape
tensor_new = tensor.reshape(1,8)
tensor_new.shape
x = tensor_new.view(2,4)
x[0][3] = 100
x,tensor_new

# Stacking
tensor_1 = torch.arange(0,11)
tensor_2 = torch.arange(10,21)
tensor_3 = torch.arange(20,31)
tensor_stack = torch.stack([tensor_1,tensor_2,tensor_3],dim = 1)
tensor_stack
```
